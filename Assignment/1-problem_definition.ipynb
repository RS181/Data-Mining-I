# Problem Definition

In this notebook we are going to understand **what the model is supposed to predict or classify**, ensuring the ML task matches the actual objective.

## 1. Identify the Business Goal or Research Purpose

The main research goal is to **identify and rank the variables (features) that are most predictive of cardiac pathologies in children**. By understanding these relationships, the researchers aim to build automated classifiers that can help less experienced medical professionals accurately distinguish between healthy children and those with heart diseases. This is intended to facilitate earlier treatment, improve patient outcomes, and reduce healthcare costs.

## 2. Decide Whether the Task is Regression, Classification, Clustering, or Forecasting

The primary task is ```classification```. Specifically:

* **Supervised Learning (Classification):** The work focuses on generating predictive models to distinguish between two classes: **Normal** (absence of pathology) and **Anormal** (presence of pathology).

* **Feature Importance/Selection:** A secondary but crucial task is ranking features using methods like Mutual Information, Chi-squared tests, and Random Forest Gini importance to understand which variables most influence the classification.

* **Exploratory Tasks: **The assignment also mentions using clustering (unsupervised learning) during multivariate analysis to see if patients form similar groups.

## Determine the Type of Data Required and Expected Output

* **Input Data:** The dataset consists of anonymized records for ```17873 children```. It includes:

    * **Demographic/Physical features:** Age, weight, height, gender, and Body Mass Index (BMI).

    * **Physiological/Clinical features:** Systolic and diastolic blood pressure (SBP/DBP), heart rate (FC), heart sounds (B2), and the presence of murmurs (SOPRO).

    * **Clinical History:** Reasons for consultation and history of previous diseases.


* **Expected Output:**

    * **A Classification Label:** Predicting whether a case is ```"Normal"``` or ```"Anormal"```.

    * **Performance Metrics:** Models are evaluated based on **Correctly Classified Instances (CCI)**, **sensitivity (recall)**, **specificity**, and **AUROC**.

    * **Ranked Feature List:** A list showing the importance of variables, such as "Murmur" being the most significant predictor.
