1. In a given application, we have information about the ages of a set of
12 people. Their values are 12, 30, 24, 10, 10, 23, 43, 67, 79,
34, 56, 51.

a) What is the median of these ages? Explain. <br>
b) What is the mode of these ages? Explain.<br>
c) How would you obtain the difference between the 99% percentile
and the 10% of this set of ages, in Python? How about the 1%
percentile?<br>
d) What are the results of normalizing and standardizing these data? <br>



```python
# Enable autocomplete
%config IPCompleter.greedy=True
```

```python
import pandas as pd 

people_ages = [12, 30, 24, 10, 10, 23, 43, 67, 79,34, 56, 51]
people_ser = pd.Series(data=people_ages,dtype=int)
people_mean = people_ser.mean()
people_mode = people_ser.mode()

#a) 
print("Mean of people: ",people_mean)
#b) 
print("Mode of people: ",people_mode)
```

```python
# c) 
percentil_99 = people_ser.describe(percentiles=[0.99])['99%']
percentil_10 = people_ser.describe(percentiles=[0.10])['10%']
percentil_1 = people_ser.describe(percentiles=[0.01])['1%']
print(f"Diference between 99% percentil and 10% percentil [{percentil_99} - {percentil_10}] = ", percentil_99-percentil_10)
print(f"Diference between 99% percentil and 1% percentil [{percentil_99} - {percentil_1}] = ", percentil_99-percentil_1)
```

```python
#d) 
X_min = people_ser.min() 
X_max = people_ser.max() 

people_normalized = (people_ser - X_min) / (X_max - X_min)
print("Normalized people data:\n",people_normalized)

# y = (x - mean) / standard_deviation
people_mean = people_ser.mean()
people_std = people_ser.std()
people_standardized = (people_ser - people_mean)/people_std
print("Standardized people data:\n",people_standardized)
print(people_standardized.std())
```
