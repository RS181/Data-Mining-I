# **ICD/FDS Regression**

Download the [Melbourne housing dataset](https://www.kaggle.com/datasets/saadmehar/melbourne-housing-fullcsv?select=Melbourne_housing_FULL.csv) and let's study housing prices and other variables. Our task is supervised and the **target variable** is the `Price`. Your job is to build a regression model that can approximate the price of new houses.

````python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
import pandas as pd

houses_full=pd.read_csv('praticas/DataSets/Melbourne_housing_FULL.csv')
houses=houses_full.dropna()
houses=houses.loc[houses['BuildingArea']<1500]
houses=houses.sample(1000,random_state=123)

houses.head()
````

Prepare for modeling...

````python
# separate the predictors as x
# transform the data frame into an numpy array
# this will facilitate slicing for plotting and ML
predictors=['BuildingArea','Rooms','Distance']
x=np.array(houses.loc[:,predictors])
y=np.array(houses.loc[:,'Price'])

def priceshrink(v,t):
    return int(v/1000000)

fig, sc = plt.subplots(1,len(predictors), figsize=(16,5))
fig.suptitle('Scatter plots between predictors and target')
for i in range(len(predictors)):
    sc[i].scatter(x[:,i], y)
    sc[i].set(xlabel=predictors[i])
    sc[i].yaxis.set_major_formatter(plt.NullFormatter())
sc[0].set(ylabel='Price')
sc[0].yaxis.set_major_formatter(plt.FuncFormatter(priceshrink))
np.shape(y)
````

* Quando fazemos o plot de **Predictor Variables** e **Target Variable**, estamos a tentar verificar se:
    1. **Existe uma relação?** (```Se o X muda, o Y muda de forma previsível?```)

    2. **Qual é a direção da relação?** (Correlação Positiva ou Negativa?)

    3. **Quão proximos estão os pontos?** (**Correlação Forte ou Fraca**)

    4. **Qual é a forma da relação?** (Linear, curva/exponencial, ou aleatória?)
* Observações da figura acima:
    * ```BuildingArea vs. Price```
        * Existe uma **correlação positiva**. 
        * **À medida que a área de construção (BuildingArea) aumenta, o preço (Price) tende a subir**. 
    * ```Rooms vs. Price``` 
        * Existe uma **correlação positiva**, mas a variável é discreta (números inteiros). 
        * É por isso que vemos linhas verticais em vez de uma nuvem dispersa.  
        * **Casas com mais quartos tendem a ser mais caras**
    * ```Distance vs. Price```
        * Existe uma **correlação negativa**. 
        * **À medida que a distância (Distance) aumenta (provavelmente distância ao centro da cidade ou praia), o preço (Price) diminui.**
* ```Qual deles das variáveis devemos usar no Modelo de Simple Linear Regression??```
    * R: Em **Simple Linear regression** Devemos escolher um **Predictor Variable** que tenha ```Correlação Forte```, e neste caso parece que **Building Area** é uma **Predictor Variable** boa (ver a tabela abaixo com algumas indicações)

| O que vemos no gráfico    | O que significa (Regressão) | Veredito |
| :-------- | ------- | ------- |
| **Pontos muito juntos** (Tubo fino) | ```O modelo tem certeza.``` <br> O erro será baixo. | **Excelente** 
| **Pontos espalhados** (Nuvem gorda) | ```O modelo tem dúvidas```. <br> O erro será alto. | **Fraco** 
| **Pontos formam uma Reta**  | A matemática da regressão (y=mx+b) ```encaixa perfeitamente```. | **Ideal**
| Pontos formam uma Curva | A **reta** vai passar **"mais ou menos"** no meio, mas vai errar nas pontas." |  **Bom, mas requer atenção**


> Posso também utilizar métricas como o R^2  (aos modelos que utilizam uma dada predictor) e verificar qual dessas predictor teve um modelo com melhor R^2




````python
model = LinearRegression(fit_intercept=True)


model.fit(x[:,0:1], y)

xfit = np.linspace(0, max(x[:,0]), 1000)
# newaxis gives xfit the shape of a matrix
yfit = model.predict(xfit[:,np.newaxis])

# the building of the plot starts by defining an axes
# then you put the scatter plot on the axes
# you set a number of things on the axes
# etc.
# the build of the plot is axes centered
ax=plt.axes(label='lrfit')
ax.scatter(x[:,0], y)
ax.set_ylabel('Price')
ax.set_xlabel('BuildingArea')
ax.set_xlim(left=0)
ax.set_xlim(left=0,right=max(x[:,0]))
ax.yaxis.set_major_formatter(plt.FuncFormatter(priceshrink))
ax.plot(xfit, yfit,color='red')
````

````python
print("Model slope:    ", model.coef_[0])
print("Model intercept:", model.intercept_)
````

````python
print("Model R2:    ", model.score(x[:,0:1],y))
````
