### Handling Outliers

An Outlier is a data-item/object that deviates significantly from the rest of the (so-called normal)objects. They can be caused by measurement or execution errors. The analysis for outlier detection is referred to as outlier mining. There are many ways to detect the outliers, and the removal process is the data frame same as removing a data item from the panda’s dataframe.

Let's consider the iris dataset and let's plot the boxplot for the SepalWidthCm column.

````python
# importing packages
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('DataSets/Iris.csv')

sns.boxplot(x='SepalWidthCm', data=df)
````

In the above graph, the values above 4 and below 2 are acting as outliers

#### Removing Outliers

For removing the outlier, one must follow the same process of removing an entry from the dataset using its exact position in the dataset because in all the above methods of detecting the outliers end result is the list of all those data items that satisfy the outlier definition according to the method used.

Example: We will detect the outliers using IQR and then we will remove them. We will also draw the boxplot to see if the outliers are removed or not.

````python
# Importing
import numpy as np

# Load the dataset 
df = pd.read_csv('DataSets/Iris.csv') 

# IQR 
Q1 = np.percentile(df['SepalWidthCm'], 25, 
				interpolation = 'midpoint') 

Q3 = np.percentile(df['SepalWidthCm'], 75, 
				interpolation = 'midpoint') 
IQR = Q3 - Q1 

print("Old Shape: ", df.shape) 

# Upper bound 
upper = np.where(df['SepalWidthCm'] >= (Q3+1.5*IQR)) 

# Lower bound 
lower = np.where(df['SepalWidthCm'] <= (Q1-1.5*IQR)) 

# Removing the Outliers 
df.drop(upper[0], inplace = True) 
df.drop(lower[0], inplace = True) 

print("New Shape: ", df.shape) 

sns.boxplot(x='SepalWidthCm', data=df)
````

4. Use the iris data set and randomly delete 15 values from the first column and pretend they are missing data. Consider different imputation approaches to 'guess' the missing values. Calculate the Mean Average Error of each approach by using the hidden values and the imputed values. Try the approaches:

    * Replace by the mean of the column.
    * Replace by the mean of the respective class.
    * Replace by the value of the nearest neighbor.
    


````python
import pandas as pd 
import numpy as np 
from sklearn.metrics import mean_absolute_error 
df = pd.read_csv("DataSets/Iris.csv")

# Put 15 data objects equal to NAN 
# Obtém todas as colunas exceto 'Species' e 'Id' e coloca-as = NAN
n_missing = 15 
missing_indices = df.sample(n=n_missing, random_state=42).index
df_aux = df.copy()
feature_col = df_aux['Species'].tolist()
id_col = df_aux['Id'].tolist()
df_aux = df_aux.drop('Id',axis=1)
df_aux = df_aux.drop('Species',axis=1)

first_column_name = df_aux.columns[0] # a.k.a SepalLengthCm
df_aux.loc[missing_indices,first_column_name] = np.nan

print("Indices that are going to be have missing values on the column 'SepalLengthCM' : ", missing_indices)
````

### Replace by the mean of the column.

````python

# Fazer fill dos missing values com a média da respetiva coluna 
mean_fill = df_aux.copy()
for column in mean_fill.columns:
    mean = mean_fill[column].mean()
    mean_fill[column] = mean_fill[column].fillna(mean)

hidden_values = df.loc[missing_indices].copy().drop(['Id','Species'],axis=1)
imputed_values = mean_fill.loc[missing_indices]

# Calcular o MAE total
mae_total = mean_absolute_error(hidden_values, imputed_values)
# Calcular o MAE por coluna
mae_per_column = (hidden_values - imputed_values).abs().mean()

print("\n" + "="*30)
print("Cálculo do Erro (Mean Absolute Error)")
print(f"MAE Total (Erro médio geral): {mae_total:.4f}")
print("\nMAE por Coluna (Erro médio por característica):")
print(mae_per_column)
print("="*30)
````

### Explicação do Medium Absolute Error (a.k.a Mean Average Error)

1. **MAE Total (Erro Médio Absoluto Total)**

    O **MAE Total** é a **"nota final" do seu método de imputação**.

    * **O que é**:
        *  É a média de todos os erros absolutos, calculados a partir de todas as células que você preencheu (neste caso, 15 linhas $\times$ 4 colunas = 60 valores).
    * **O que indica**:
        * Dá-lhe um único número que representa o erro médio geral do seu método.
    * **Como interpretar**:
        * Se o seu MAE Total for $0.45$, isso significa que, em média, cada valor que o seu método "adivinhou" estava a $0.45$ unidades de distância do valor real (ex: $0.45$ cm).
    * **Principal Uso**: Comparar métodos diferentes.
        * Método A (Média da Coluna): MAE Total = $0.45$
        * Método B (Média da Classe): MAE Total = $0.25$
        * Método C (KNN): MAE Total = $0.15$
        * **Conclusão**: O **Método C (KNN) é o melhor**, **porque o seu erro médio geral foi o mais baixo**.

2. **MAE por Coluna (Erro Médio Absoluto por Coluna)**

    O **MAE por Coluna** é o **"diagnóstico" ou o "boletim de notas" detalhado**.

   * **O que é**:
       * É o erro médio calculado especificamente para cada coluna. (Ex: a média dos 15 erros da 'SepalLengthCm', a média dos 15 erros da 'PetalWidthCm', etc.).
   * **O que indica**:
       * Mostra-lhe onde (em que características) o seu método de imputação está a ser preciso ou impreciso.
   * **Como interpretar**:
       * Imagine que o seu MAE Total foi $0.45$, mas o MAE por Coluna foi :
           * SepalLengthCm: $0.95$
           * SepalWidthCm: $0.60$
           * PetalLengthCm: $0.15$
           *  PetalWidthCm: $0.10$
   * **Principal Uso**: **Diagnosticar o seu método**.
       * **Conclusão(dos valores do ponto anterior)**: A sua imputação (ex: Média da Coluna) está a "adivinhar" muito bem os valores de PetalLengthCm e PetalWidthCm (erros baixos). No entanto, está a falhar redondamente em "adivinhar" a SepalLengthCm (erro muito alto).
       * Isto pode indicar que a SepalLengthCm tem uma relação complexa com outras variáveis (como a Species) que um método simples (como a média da coluna) não consegue capturar.
